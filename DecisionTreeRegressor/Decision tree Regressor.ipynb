```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/Users/thotaruchitha/Downloads/Data.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train , y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
#To perform feature scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
y_train = sc.fit_transform(y_train.reshape(-1, 1))
```

```python
y_train
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
Regressor = DecisionTreeRegressor()
Regressor.fit(x_train,y_train)
```

```python
y_pred = Regressor.predict(x_test)
```

```python
y_pred
```

```python
# to invert the standardization transformation applied to the predicted values
y_pred = sc.inverse_transform(y_pred.reshape(-1,1))
y_pred
```

```python
from sklearn.metrics import r2_score
print(f'The r2 score is : {r2_score(y_test ,y_pred)}')
```
