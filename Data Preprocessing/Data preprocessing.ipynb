```python
# Extracting independent variables
x=df.iloc[:,:-1].values
```

```python
# Extracting dependent variables
y=df.iloc[:,-1].values
```

```python
y
```

```python
#handling missing values
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(missing_values = np.nan , strategy = 'mean')
x[:,1:3] = imputer.fit_transform(x[:,1:3])
```

```python
x
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Create a ColumnTransformer to apply OneHotEncoder to the first column
ct = ColumnTransformer(
    transformers=[('encode', OneHotEncoder(), [0])],
    remainder='passthrough'  # Keep the other columns as is
)

# Transform the data
x = ct.fit_transform(x)

```

```python
x
```

```python
# splitting into train_test _split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split (x,y,test_size =0.2,random_state = 42)
```

```python
x_train
```

```python
#To perform feature scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```
